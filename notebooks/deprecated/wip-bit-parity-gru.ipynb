📊 Generating parity dataset with binary sequences:


```python
import random
import torch
import numpy as np
import itertools
from sklearn.model_selection import train_test_split
from torch.utils.data import TensorDataset
from tsilva_notebook_utils.torch import get_current_device, inspect_tensor_dataset

def generate_parity_dataset(
    sequence_length=None,
    variable_sequence_length=None,
    val_size=None,
    seed=None
):
    if sequence_length is None: sequence_length = CONFIG['sequence_length']
    if variable_sequence_length is None: variable_sequence_length = CONFIG['variable_sequence_length']
    if val_size is None: val_size = CONFIG['val_size']
    if seed is None: seed = CONFIG['seed']

    # Set seed for reproducibility
    set_seed(seed)

    data = []
    labels = []

    # Determine the range of sequence lengths to generate
    min_sequence_length = 1
    max_sequence_length = sequence_length + 1
    if not variable_sequence_length:
        # If variable lengths are not allowed, generate only fixed-length sequences
        min_sequence_length = max_sequence_length - 1

    # Generate all binary sequences for each length in the range
    for length in range(min_sequence_length, max_sequence_length):
        sequences = list(itertools.product([0, 1], repeat=length))  # All combinations of 0s and 1s
        for seq in sequences:
            # Right-pad sequence with 0s to make them all the same length
            padded_seq = list(seq) + [0] * (sequence_length - length)
            parity = sum(seq) % 2  # Label is 1 if number of 1s is odd, else 0
            data.append(padded_seq)
            labels.append(parity)

    # Convert data and labels to PyTorch tensors on the correct device
    device = get_current_device()
    X = torch.tensor(data, dtype=torch.float32, device=device).unsqueeze(-1)  # Shape: (N, sequence_length, 1)
    Y = torch.tensor(labels, dtype=torch.long, device=device)

    # Shuffle the dataset
    perm = torch.randperm(len(X))
    X = X[perm]
    Y = Y[perm]

    # Split into training and validation datasets
    X_train, X_val, Y_train, Y_val = train_test_split(
        X, Y, test_size=val_size, random_state=seed, shuffle=True
    )

    # Create TensorDatasets
    train_dataset = TensorDataset(X_train, Y_train)
    val_dataset = TensorDataset(X_val, Y_val)

    return train_dataset, val_dataset

train_dataset, val_dataset = generate_parity_dataset()
inspect_tensor_dataset(train_dataset)
```

🧠 Let's build the GRU model:

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import random

class SimpleGRUCell(nn.Module):
    def __init__(self, input_size, hidden_size):
        super().__init__()
        self.hidden_size = hidden_size

        # Update gate weights – controls how much of the previous hidden state
        # vs. the candidate hidden state contributes to the new hidden state
        self.input_to_update = nn.Linear(input_size, hidden_size)
        self.hidden_to_update = nn.Linear(hidden_size, hidden_size)

        # Reset gate weights – controls how much of the previous hidden state
        # is used to compute the candidate hidden state
        self.input_to_reset = nn.Linear(input_size, hidden_size)
        self.hidden_to_reset = nn.Linear(hidden_size, hidden_size)

        # Candidate hidden state weights – used to compute a new candidate
        # hidden state based on the current input and (reset-modulated) hidden state
        self.input_to_candidate = nn.Linear(input_size, hidden_size)
        self.hidden_to_candidate = nn.Linear(hidden_size, hidden_size)

    def forward(self, current_input, prev_hidden):
        # current_input: (batch_size, input_size)
        # prev_hidden: (batch_size, hidden_size)

        # 1. Compute update gate
        update_gate = torch.sigmoid(
            self.input_to_update(current_input) +
            self.hidden_to_update(prev_hidden)
        )

        # 2. Compute reset gate
        reset_gate = torch.sigmoid(
            self.input_to_reset(current_input) +
            self.hidden_to_reset(prev_hidden)
        )

        # 3. Compute candidate hidden state
        reset_hidden = reset_gate * prev_hidden
        candidate_hidden = torch.tanh(
            self.input_to_candidate(current_input) +
            self.hidden_to_candidate(reset_hidden)
        )

        # 4. Compute new hidden state
        new_hidden = (1 - update_gate) * prev_hidden + update_gate * candidate_hidden

        return new_hidden

class SimpleGRU(nn.Module):
    def __init__(self, input_size, hidden_size):
        super().__init__()
        self.hidden_size = hidden_size
        self.cell = SimpleGRUCell(input_size, hidden_size)

    def forward(self, x, h_0=None):
        batch_size, seq_len, _ = x.size()

        if h_0 is None:
            h_t = torch.zeros(batch_size, self.hidden_size, device=x.device)
        else:
            h_t = h_0.squeeze(0)

        outputs = []

        for t in range(seq_len):
            h_t = self.cell(x[:, t, :], h_t)
            outputs.append(h_t.unsqueeze(1))  # (batch, 1, hidden)

        output = torch.cat(outputs, dim=1)           # (batch, seq_len, hidden)
        h_n = h_t.unsqueeze(0)                        # (1, batch, hidden)

        return output, h_n
```

Define the GRUClassifier architecture.

```python
class GRUClassifier(nn.Module):
    def __init__(self, input_size, hidden_size):
        super().__init__()
        self.gru = SimpleGRU(input_size, hidden_size)
        self.fc = nn.Linear(hidden_size, 1)

    def forward(self, x):
        _, h_n = self.gru(x)
        h_n = h_n.squeeze(0)  # (batch, hidden)
        out = self.fc(h_n)
        return torch.sigmoid(out).squeeze(1)

input_size = 1
hidden_size = 16
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = GRUClassifier(input_size, hidden_size).to(device)
```

📝 Evaluating model performance before training:


```python
# Define the loss function as Cross Entropy Loss, typically used for classification tasks
loss_fn = nn.CrossEntropyLoss()

# Internal evaluation function that computes loss and accuracy, and optionally collects misclassified examples
def _eval(loader, collect_misses=False):
    losses = []      # Store losses for each batch
    accuracies = []  # Store accuracy for each batch
    misses = []      # Store misclassified examples if requested

    for x_batch, y_batch in loader:
        # Disable gradient calculation for evaluation (saves memory and computation)
        with torch.no_grad():
            logits, _ = model(x_batch)  # Forward pass to get predictions

        loss = loss_fn(logits, y_batch)  # Compute the loss for the batch
        losses.append(loss.item())       # Save scalar loss value

        predicted = torch.argmax(logits, dim=1)  # Convert logits to class predictions
        accuracy = (predicted == y_batch).float().mean().item()  # Calculate accuracy
        accuracies.append(accuracy)  # Save accuracy

        # If we're not collecting misclassified examples, skip the next steps
        if not collect_misses:
            continue

        # Identify which predictions are incorrect
        mismatches = predicted != y_batch
        if not mismatches.any():
            continue  # If all predictions are correct, move to next batch

        # Extract the misclassified inputs and their true/predicted labels
        failed_x = x_batch[mismatches]
        failed_y_true = y_batch[mismatches]
        failed_y_pred = predicted[mismatches]

        # Collect the details of each misclassified example
        for x, y_true, y_pred in zip(failed_x, failed_y_true, failed_y_pred):
            misses.append({
                'input': x.cpu(),                # Store input (moved to CPU)
                'true_label': y_true.item(),     # Actual label
                'predicted_label': y_pred.item() # Predicted label
            })

    # Compute average loss and accuracy over all batches
    avg_loss = sum(losses) / len(loader)
    avg_accuracy = sum(accuracies) / len(accuracies)

    results = {
        'loss': avg_loss,
        'accuracy': avg_accuracy
    }

    # If misses were collected, add them to the results
    if collect_misses:
        results['misses'] = misses

    return results

# Public evaluation function that temporarily sets model to eval mode, runs evaluation, then resets to train mode
def eval(loader, collect_misses=False):
    model.eval()  # Set model to evaluation mode (e.g., disables dropout, batchnorm tracking)
    try:
        return _eval(loader, collect_misses=collect_misses)  # Run evaluation
    finally:
        model.train()  # Return model to training mode

# Run evaluation on both training and validation data without collecting misclassified examples
{
    "train": eval(train_loader, collect_misses=False),
    "val": eval(val_loader, collect_misses=False),
}
```

🚀 Training the model with early stopping at perfect accuracy:

```python
import torch.optim as optim
from torch.nn.utils import clip_grad_norm_
from tqdm import tqdm
import wandb
from tsilva_notebook_utils.torch import calc_grad_norm

def train(
    n_epochs=None,
    learning_rate=None,
    max_grad_norm=None,
    weight_decay=None,
    warmup_ratio=None
):
    # Use defaults from CONFIG if not specified
    if n_epochs is None: n_epochs = CONFIG['n_epochs']
    if learning_rate is None: learning_rate = CONFIG['learning_rate']
    if max_grad_norm is None: max_grad_norm = CONFIG['max_grad_norm']
    if weight_decay is None: weight_decay = CONFIG['weight_decay']
    if warmup_ratio is None: warmup_ratio = CONFIG['warmup_ratio']

    model.train()  # Set model to training mode
    wandb.watch(model, log="all")  # Watch model for gradient logging with wandb

    # Set up optimizer and learning rate scheduler
    optimizer = optim.Adam(model.parameters(), lr=learning_rate, weight_decay=weight_decay)
    total_steps = n_epochs * len(train_loader)
    warmup_steps = int(warmup_ratio * total_steps)

    # Linear warmup schedule: linearly increase LR for warmup_steps, then keep constant
    def lr_lambda(current_step):
        return float(current_step) / float(max(1, warmup_steps)) if current_step < warmup_steps else 1.0

    scheduler = torch.optim.lr_scheduler.LambdaLR(optimizer, lr_lambda=lr_lambda)
    global_step = 0

    # Progress bar for training
    with tqdm(range(n_epochs), desc="Training") as pbar:
        for epoch in pbar:
            losses, accuracies, grad_norms = [], [], []
            last_layer_norms = {}

            # Iterate over batches
            for x_batch, y_batch in train_loader:
                global_step += 1

                logits, _ = model(x_batch)
                loss = loss_fn(logits, y_batch)
                losses.append(loss.item())

                optimizer.zero_grad()
                loss.backward()

                # Optionally clip gradients to avoid exploding gradients
                if max_grad_norm > 0:
                    clip_grad_norm_(model.parameters(), max_grad_norm)

                # Calculate gradient norms for monitoring
                grad_norm, layer_norms = calc_grad_norm(model)
                grad_norms.append(grad_norm)
                last_layer_norms = layer_norms

                optimizer.step()
                scheduler.step()

                # Calculate training accuracy
                with torch.no_grad():
                    predicted = torch.argmax(logits, dim=1)
                    accuracy = (predicted == y_batch).float().mean().item()
                    accuracies.append(accuracy)

            # Aggregate stats for the epoch
            train_loss = sum(losses) / len(losses)
            train_accuracy = sum(accuracies) / len(accuracies)
            train_grad_norm = sum(grad_norms) / len(grad_norms)

            # Run evaluation on validation set
            result = eval(val_loader)
            val_loss = result['loss']
            val_accuracy = result['accuracy']
            current_lr = scheduler.get_last_lr()[0]

            # Structure logs for W&B dashboard
            stats = {
                'meta/epoch': epoch + 1,
                'meta/step': global_step,
                'meta/learning_rate': current_lr,

                'train/loss': train_loss,
                'train/accuracy': train_accuracy,
                'grad/total_norm': train_grad_norm,

                'val/loss': val_loss,
                'val/accuracy': val_accuracy,
            }

            # Log detailed gradient norms for select layers on final epoch
            if epoch == n_epochs - 1:
                for layer_name, norm_val in last_layer_norms.items():
                    if "output" in layer_name or "classifier" in layer_name:
                        stats[f'grad/layer/{layer_name}'] = norm_val

            wandb.log(stats, step=global_step)  # Log all stats to W&B

            # Show key metrics in progress bar
            pbar.set_postfix({
                'epoch': epoch + 1,
                'lr': f'{current_lr:.6e}',
                'grad_norm': f'{train_grad_norm:.2f}',
                'train_acc': f'{train_accuracy * 100:.2f}%',
                'val_acc': f'{val_accuracy * 100:.2f}%'
            })

            # Early stopping if training accuracy hits 100%
            if train_accuracy == 1.0:
                print("\nTraining finished early: model perfectly fit the training set.")
                break

    wandb.finish()

# Call training
train()
```

🔍 Evaluating final model and analyzing missed predictions:


```python
def log_misses(misses):
    for i, miss in enumerate(misses):
        input = miss["input"]
        input_s = "".join([str(x.item()) for x in input.int().squeeze()])
        correct = "odd" if miss["true_label"] == 1 else "even"
        predicted = "odd" if miss["predicted_label"] == 1 else "even"
        print(f"{input_s}: {correct} (predicted: {predicted})")

result = eval(val_loader, collect_misses=True)
val_accuracy = result['accuracy']
misses = result['misses']
print(f"Validation accuracy: {val_accuracy * 100:.2f}%")
log_misses(misses)
```
