# Visual Guide to Tensor Multiplication

This notebook builds your intuition for tensor (matrix) multiplication through progressive visualizations. By the end, you'll be able to:

- Visualize how matrices are multiplied
- Perform small multiplications in your head
- Understand the patterns that make it easier

**Key Insight**: Matrix multiplication is all about **dot products between rows and columns**.

## Setup

We'll use PyTorch for tensor operations and matplotlib/seaborn for visualizations.

````python
import torch
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.patches import Rectangle, FancyBboxPatch
from matplotlib.colors import LinearSegmentedColormap

# Set style for better visuals
sns.set_style('white')
plt.rcParams['figure.figsize'] = (12, 6)
plt.rcParams['font.size'] = 11
````

## 1. The Building Block: Dot Product

Everything starts with the **dot product** of two vectors:

$$\text{dot}(a, b) = a_1 \cdot b_1 + a_2 \cdot b_2 + \ldots + a_n \cdot b_n$$

**Mental Model**: Pair up elements, multiply them, then sum everything.

````python
# Simple dot product example
a = torch.tensor([2., 3., 1.])
b = torch.tensor([1., 4., 2.])

dot_product = torch.dot(a, b)
print(f"a = {a.numpy()}")
print(f"b = {b.numpy()}")
print(f"\ndot(a, b) = {dot_product.item()}")
````

### Visualizing the Dot Product

Let's see this step-by-step: multiply corresponding elements, then add them all up.

````python
fig, axes = plt.subplots(1, 3, figsize=(14, 3))

# Vector a
ax = axes[0]
colors_a = ['#FF6B6B', '#4ECDC4', '#45B7D1']
ax.bar(range(len(a)), a.numpy(), color=colors_a, alpha=0.7, edgecolor='black')
ax.set_ylim(0, 5)
ax.set_xticks(range(len(a)))
ax.set_title('Vector a', fontsize=14, fontweight='bold')
ax.set_ylabel('Value')
for i, val in enumerate(a):
    ax.text(i, val + 0.1, f'{val:.0f}', ha='center', va='bottom', fontweight='bold')

# Vector b
ax = axes[1]
ax.bar(range(len(b)), b.numpy(), color=colors_a, alpha=0.7, edgecolor='black')
ax.set_ylim(0, 5)
ax.set_xticks(range(len(b)))
ax.set_title('Vector b', fontsize=14, fontweight='bold')
for i, val in enumerate(b):
    ax.text(i, val + 0.1, f'{val:.0f}', ha='center', va='bottom', fontweight='bold')

# Element-wise products
ax = axes[2]
products = a * b
bars = ax.bar(range(len(products)), products.numpy(), color=colors_a, alpha=0.7, edgecolor='black')
ax.set_ylim(0, 15)
ax.set_xticks(range(len(products)))
ax.set_title('Element-wise Products (sum = dot product)', fontsize=14, fontweight='bold')
for i, val in enumerate(products):
    ax.text(i, val + 0.3, f'{a[i]:.0f}×{b[i]:.0f}={val:.0f}', ha='center', va='bottom', fontweight='bold')

# Add sum line
ax.axhline(y=dot_product.item(), color='red', linestyle='--', linewidth=2, label=f'Sum = {dot_product:.0f}')
ax.legend()

plt.tight_layout()
plt.show()

print(f"\nStep-by-step: 2×1 + 3×4 + 1×2 = 2 + 12 + 2 = 16")
````

## 2. Matrix-Vector Multiplication

Now the key leap: **Matrix × Vector = Vector of dot products**

Each row of the matrix takes a dot product with the vector.

$$\begin{bmatrix} \text{row 1} \\ \text{row 2} \\ \text{row 3} \end{bmatrix} \times \begin{bmatrix} v \end{bmatrix} = \begin{bmatrix} \text{row 1} \cdot v \\ \text{row 2} \cdot v \\ \text{row 3} \cdot v \end{bmatrix}$$

**Mental Model**: Each output element = one row × the vector.

````python
# Matrix-vector example
M = torch.tensor([
    [2., 1., 3.],
    [1., 0., 2.],
    [3., 2., 1.]
])

v = torch.tensor([1., 2., 1.])

result = M @ v  # @ is matrix multiplication

print("Matrix M:")
print(M.numpy())
print(f"\nVector v: {v.numpy()}")
print(f"\nResult M @ v: {result.numpy()}")
````

### Visualizing Matrix-Vector Multiplication

Watch how each row of M combines with vector v to produce one output value.

````python
fig, axes = plt.subplots(1, 4, figsize=(16, 4))

# Show the matrix
ax = axes[0]
sns.heatmap(M.numpy(), annot=True, fmt='.0f', cmap='YlOrRd', cbar=False, 
            square=True, linewidths=2, ax=ax, vmin=0, vmax=3)
ax.set_title('Matrix M\n(3×3)', fontsize=12, fontweight='bold')
ax.set_ylabel('Rows')
ax.set_xlabel('Columns')

# Show the vector
ax = axes[1]
v_reshaped = v.numpy().reshape(-1, 1)
sns.heatmap(v_reshaped, annot=True, fmt='.0f', cmap='Blues', cbar=False,
            square=True, linewidths=2, ax=ax, vmin=0, vmax=3)
ax.set_title('Vector v\n(3×1)', fontsize=12, fontweight='bold')
ax.set_xlabel('')
ax.set_xticks([])

# Multiplication symbol
ax = axes[2]
ax.text(0.5, 0.5, '×', fontsize=60, ha='center', va='center')
ax.set_xlim(0, 1)
ax.set_ylim(0, 1)
ax.axis('off')

# Show the result
ax = axes[3]
result_reshaped = result.numpy().reshape(-1, 1)
sns.heatmap(result_reshaped, annot=True, fmt='.0f', cmap='Greens', cbar=False,
            square=True, linewidths=2, ax=ax, vmin=0, vmax=10)
ax.set_title('Result\n(3×1)', fontsize=12, fontweight='bold')
ax.set_xlabel('')
ax.set_xticks([])

plt.tight_layout()
plt.show()

# Show calculations
print("\nDetailed calculations:")
for i in range(len(M)):
    row = M[i]
    dot = torch.dot(row, v)
    calc = ' + '.join([f'{row[j]:.0f}×{v[j]:.0f}' for j in range(len(v))])
    print(f"Row {i}: [{', '.join([f'{x:.0f}' for x in row])}] · v = {calc} = {dot:.0f}")
````

### Step-by-Step: Computing One Output Element

Let's zoom in on computing the **first element** of the result. This is the pattern you'll use mentally.

````python
fig, ax = plt.subplots(1, 1, figsize=(10, 6))

# Create a visual representation
row_idx = 0
row = M[row_idx]

y_positions = [3, 2, 1]
colors = ['#FF6B6B', '#4ECDC4', '#45B7D1']

for i, (m_val, v_val) in enumerate(zip(row, v)):
    y = y_positions[i]
    
    # Matrix element
    rect1 = FancyBboxPatch((0, y-0.3), 1, 0.6, boxstyle="round,pad=0.05", 
                          edgecolor='black', facecolor=colors[i], alpha=0.6, linewidth=2)
    ax.add_patch(rect1)
    ax.text(0.5, y, f'M[0,{i}]\n{m_val:.0f}', ha='center', va='center', 
            fontweight='bold', fontsize=12)
    
    # Multiplication sign
    ax.text(1.5, y, '×', ha='center', va='center', fontsize=20, fontweight='bold')
    
    # Vector element
    rect2 = FancyBboxPatch((2, y-0.3), 1, 0.6, boxstyle="round,pad=0.05",
                          edgecolor='black', facecolor=colors[i], alpha=0.6, linewidth=2)
    ax.add_patch(rect2)
    ax.text(2.5, y, f'v[{i}]\n{v_val:.0f}', ha='center', va='center', 
            fontweight='bold', fontsize=12)
    
    # Equals sign
    ax.text(3.5, y, '=', ha='center', va='center', fontsize=20, fontweight='bold')
    
    # Product
    product = m_val * v_val
    rect3 = FancyBboxPatch((4, y-0.3), 1, 0.6, boxstyle="round,pad=0.05",
                          edgecolor='black', facecolor=colors[i], alpha=0.6, linewidth=2)
    ax.add_patch(rect3)
    ax.text(4.5, y, f'{product:.0f}', ha='center', va='center', 
            fontweight='bold', fontsize=14)

# Plus signs
ax.text(5.5, 2.5, '+', ha='center', va='center', fontsize=24, fontweight='bold')
ax.text(5.5, 1.5, '+', ha='center', va='center', fontsize=24, fontweight='bold')

# Final sum
total = torch.dot(row, v)
rect_final = FancyBboxPatch((6, 1.5), 1.5, 1, boxstyle="round,pad=0.1",
                           edgecolor='red', facecolor='lightgreen', alpha=0.7, linewidth=3)
ax.add_patch(rect_final)
ax.text(6.75, 2, f'Result[0]\n{total:.0f}', ha='center', va='center', 
        fontweight='bold', fontsize=14)

ax.set_xlim(-0.5, 8)
ax.set_ylim(0, 4)
ax.axis('off')
ax.set_title('Computing Result[0] = Row 0 · Vector', fontsize=14, fontweight='bold', pad=20)

plt.tight_layout()
plt.show()
````

## 3. Matrix-Matrix Multiplication: The Full Picture

**Key Insight**: Matrix multiplication is just **many matrix-vector multiplications** side by side!

$$\begin{bmatrix} \text{row 1} \\ \text{row 2} \end{bmatrix} \times \begin{bmatrix} \text{col 1} & \text{col 2} \end{bmatrix} = \begin{bmatrix} r1·c1 & r1·c2 \\ r2·c1 & r2·c2 \end{bmatrix}$$

**Mental Model**: Result[i, j] = Row i of A · Column j of B

````python
# Simple matrix multiplication example
A = torch.tensor([
    [1., 2.],
    [3., 4.]
])

B = torch.tensor([
    [2., 0.],
    [1., 3.]
])

C = A @ B

print("Matrix A:")
print(A.numpy())
print("\nMatrix B:")
print(B.numpy())
print("\nResult C = A @ B:")
print(C.numpy())
````

### Visualizing the Shape Rule

**Critical rule for shapes**: (m × n) @ (n × p) → (m × p)

The inner dimensions must match, and they "disappear" in the result.

````python
fig, ax = plt.subplots(1, 1, figsize=(12, 3))

# A shape
rect_a = FancyBboxPatch((0, 0.5), 2, 1, boxstyle="round,pad=0.1",
                        edgecolor='black', facecolor='#FFB6C1', alpha=0.7, linewidth=2)
ax.add_patch(rect_a)
ax.text(1, 1, 'A\n(2 × 2)', ha='center', va='center', fontweight='bold', fontsize=14)

# Multiplication sign
ax.text(2.5, 1, '@', ha='center', va='center', fontsize=24, fontweight='bold')

# B shape
rect_b = FancyBboxPatch((3, 0.5), 2, 1, boxstyle="round,pad=0.1",
                        edgecolor='black', facecolor='#ADD8E6', alpha=0.7, linewidth=2)
ax.add_patch(rect_b)
ax.text(4, 1, 'B\n(2 × 2)', ha='center', va='center', fontweight='bold', fontsize=14)

# Equals sign
ax.text(5.5, 1, '=', ha='center', va='center', fontsize=24, fontweight='bold')

# C shape
rect_c = FancyBboxPatch((6, 0.5), 2, 1, boxstyle="round,pad=0.1",
                        edgecolor='red', facecolor='#90EE90', alpha=0.7, linewidth=3)
ax.add_patch(rect_c)
ax.text(7, 1, 'C\n(2 × 2)', ha='center', va='center', fontweight='bold', fontsize=14)

# Annotations for dimensions
ax.annotate('', xy=(1, 0.3), xytext=(0.3, 0.3),
            arrowprops=dict(arrowstyle='<->', color='red', lw=2))
ax.text(0.65, 0.05, 'm=2', ha='center', fontsize=11, color='red', fontweight='bold')

ax.annotate('', xy=(1.7, 1.7), xytext=(0.3, 1.7),
            arrowprops=dict(arrowstyle='<->', color='blue', lw=2))
ax.text(1, 1.95, 'n=2', ha='center', fontsize=11, color='blue', fontweight='bold')

ax.annotate('', xy=(4.7, 1.7), xytext=(3.3, 1.7),
            arrowprops=dict(arrowstyle='<->', color='blue', lw=2))
ax.text(4, 1.95, 'n=2', ha='center', fontsize=11, color='blue', fontweight='bold')

ax.annotate('', xy=(7.7, 1.7), xytext=(6.3, 1.7),
            arrowprops=dict(arrowstyle='<->', color='green', lw=2))
ax.text(7, 1.95, 'p=2', ha='center', fontsize=11, color='green', fontweight='bold')

ax.set_xlim(-0.5, 8.5)
ax.set_ylim(-0.2, 2.3)
ax.axis('off')
ax.set_title('Shape Rule: (m × n) @ (n × p) → (m × p)', fontsize=14, fontweight='bold', pad=10)

plt.tight_layout()
plt.show()

print("✓ Inner dimensions (n) must match")
print("✓ Outer dimensions (m, p) determine result shape")
````

## 4. Mental Math Strategy

To do small matrix multiplications in your head, follow this pattern:

1. **Identify result position**: "I want C[0, 1]"
2. **Find the row**: Row 0 of A
3. **Find the column**: Column 1 of B
4. **Compute dot product**: Pair up and multiply, then sum

Let's practice with a fresh example.

````python
# Practice example
A_practice = torch.tensor([
    [2., 3.],
    [1., 4.]
])

B_practice = torch.tensor([
    [1., 2.],
    [3., 1.]
])

print("Try to compute C = A @ B mentally!\n")
print("Matrix A:")
print(A_practice.numpy())
print("\nMatrix B:")
print(B_practice.numpy())
print("\n" + "="*50)
print("THINK: What is C[0,0]?")
print("  Row 0 of A: [2, 3]")
print("  Col 0 of B: [1, 3]")
print("  Dot product: 2×1 + 3×3 = 2 + 9 = 11")
print("\nTHINK: What is C[0,1]?")
print("  Row 0 of A: [2, 3]")
print("  Col 1 of B: [2, 1]")
print("  Dot product: 2×2 + 3×1 = 4 + 3 = 7")
print("="*50)

C_practice = A_practice @ B_practice
print("\nAnswer:")
print(C_practice.numpy())
````

### Interactive Practice Tool

Here's a helper function to quiz yourself on any position.

````python
def show_computation(A, B, i, j):
    """Show how to compute C[i,j] = A @ B step by step."""
    row = A[i]
    col = B[:, j]
    
    print(f"Computing C[{i},{j}]:")
    print(f"  Row {i} of A: {row.numpy()}")
    print(f"  Col {j} of B: {col.numpy()}")
    
    products = row * col
    print(f"\n  Element-wise: {' + '.join([f'{row[k]:.0f}×{col[k]:.0f}' for k in range(len(row))])}")
    print(f"              = {' + '.join([f'{products[k]:.0f}' for k in range(len(products))])}")
    
    result = torch.dot(row, col)
    print(f"              = {result:.0f}")
    
    return result

# Try it out!
show_computation(A_practice, B_practice, 1, 0)
print("\n" + "="*50 + "\n")
show_computation(A_practice, B_practice, 1, 1)
````

## 5. Non-Square Matrices

The same rules apply! Remember: **(m × n) @ (n × p) → (m × p)**

The inner dimension must match, outer dimensions become the result shape.

````python
# Different sized matrices
A_rect = torch.tensor([
    [1., 2., 3.],
    [4., 5., 6.]
])  # 2×3

B_rect = torch.tensor([
    [1., 2.],
    [3., 4.],
    [5., 6.]
])  # 3×2

C_rect = A_rect @ B_rect  # Should be 2×2

print(f"A shape: {A_rect.shape} (2×3)")
print(A_rect.numpy())
print(f"\nB shape: {B_rect.shape} (3×2)")
print(B_rect.numpy())
print(f"\nC shape: {C_rect.shape} (2×2)")
print(C_rect.numpy())
````

### Visualizing Non-Square Multiplication

Notice how the shapes work out perfectly.

````python
fig = plt.figure(figsize=(16, 5))

# Create grid spec for unequal sizes
gs = fig.add_gridspec(2, 7, width_ratios=[3, 0.5, 2, 0.5, 2, 0.5, 2], 
                      height_ratios=[1, 1], hspace=0.3, wspace=0.3)

# A (2×3)
ax = fig.add_subplot(gs[:, 0])
sns.heatmap(A_rect.numpy(), annot=True, fmt='.0f', cmap='Reds', cbar=False,
           linewidths=2, ax=ax, vmin=0, vmax=6)
ax.set_title('A (2×3)', fontweight='bold', fontsize=13)

# @ symbol
ax = fig.add_subplot(gs[:, 1])
ax.text(0.5, 0.5, '@', fontsize=40, ha='center', va='center', fontweight='bold')
ax.axis('off')

# B (3×2)
ax = fig.add_subplot(gs[:, 2])
sns.heatmap(B_rect.numpy(), annot=True, fmt='.0f', cmap='Blues', cbar=False,
           linewidths=2, ax=ax, vmin=0, vmax=6)
ax.set_title('B (3×2)', fontweight='bold', fontsize=13)

# = symbol
ax = fig.add_subplot(gs[:, 3])
ax.text(0.5, 0.5, '=', fontsize=40, ha='center', va='center', fontweight='bold')
ax.axis('off')

# C (2×2)
ax = fig.add_subplot(gs[:, 4])
sns.heatmap(C_rect.numpy(), annot=True, fmt='.0f', cmap='Greens', cbar=False,
           linewidths=3, ax=ax, vmin=0, vmax=50, linecolor='red')
ax.set_title('C (2×2)', fontweight='bold', fontsize=13)

plt.suptitle('Non-Square Matrix Multiplication: (2×3) @ (3×2) → (2×2)',
            fontsize=14, fontweight='bold')
plt.show()

# Show one example calculation
print("\nExample: C[0,0]")
row = A_rect[0]
col = B_rect[:, 0]
print(f"  Row 0 of A: {row.numpy()}")
print(f"  Col 0 of B: {col.numpy()}")
print(f"  Result: 1×1 + 2×3 + 3×5 = 1 + 6 + 15 = {torch.dot(row, col):.0f}")
````

## 6. Common Patterns in Deep Learning

Understanding these shapes will help you debug neural networks!

**Linear Layer**: (batch_size × input_dim) @ (input_dim × output_dim) → (batch_size × output_dim)

````python
# Simulating a batch of data through a linear layer
batch_size = 4
input_dim = 3
output_dim = 2

# Batch of inputs (e.g., 4 samples, each with 3 features)
X = torch.randn(batch_size, input_dim)

# Weight matrix of a linear layer
W = torch.randn(input_dim, output_dim)

# Forward pass
Y = X @ W

print(f"Input shape: {X.shape} (batch_size={batch_size}, input_dim={input_dim})")
print(f"Weight shape: {W.shape} (input_dim={input_dim}, output_dim={output_dim})")
print(f"Output shape: {Y.shape} (batch_size={batch_size}, output_dim={output_dim})")
print("\n✓ Each sample is transformed from 3D to 2D")
print("✓ Batch dimension is preserved")
````

### Visualizing the Batch Operation

Each row of the input batch is independently transformed by the weight matrix.

## 7. Key Takeaways & Mental Checklist

**When you see A @ B, think:**

1. **Shapes**: (m × n) @ (n × p) → (m × p)
   - Inner dimensions (n) must match
   - They "disappear" in the result

2. **Computation**: Result[i, j] = Row i of A · Column j of B
   - Each element is a dot product
   - Pair up elements, multiply, sum

3. **Mental Math Pattern**:
   - Pick your position [i, j]
   - Grab row i from A
   - Grab column j from B
   - Compute dot product

4. **Visualization**: 
   - Rows of A "sweep across" columns of B
   - Each intersection is one dot product

## 10. Summary: The Complete Mental Model

**Matrix multiplication is elegant because it's just repeated dot products!**

```
        B
      -----
    | c c c
    | o o o
A   | l l l
--- | 
r → | × × ×  ← Each × is one dot product
o → | × × ×
w → | × × ×
```

**Remember:**
- Shapes: **(m × n) @ (n × p) → (m × p)**
- Pattern: **Result[i,j] = Row i · Column j**
- Computation: **Multiply pairs, sum everything**

With practice, you'll start seeing the patterns instantly! 🎯
